```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from dotenv import load_dotenv
import os
```

```python
load_dotenv("../.env")
census_key = os.getenv("CENSUS_API_KEY")
```

```python
education_data = pd.read_csv('../data/Education/education_data.csv')
```

```python
unemployment_data = pd.read_csv('../data/Unemployment/unemployment_data.csv')
unemployment_data
```

```python
poverty_data = pd.read_csv('../data/Poverty and Income/poverty_and_income.csv')
poverty_data
```

```python
marital_data = pd.read_csv('../data/Marital Status/marital_status.csv')
marital_data
```

```python
df = pd.read_csv('../data/Medicare Part D Opioid Prescribing Rates - by Geography/2023/OMT_MDCR_RY25_P04_V10_Y23_GEO.csv')
df
```

```python
county_df = df[df['Prscrbr_Geo_Lvl'] == 'County']
county_df
```

```python
west_virginia_counties = county_df[county_df['Prscrbr_Geo_Desc'].str.contains('West Virginia')]

west_virginia_counties = west_virginia_counties[
    (west_virginia_counties['Year'] >= 2019) &
    (west_virginia_counties['Year'] <= 2023)
]

west_virginia_counties

```

```python
df_opioid = west_virginia_counties [west_virginia_counties['Breakout'] == 'Overall']

columns_to_keep = [
    'Year',
    'Prscrbr_Geo_Cd',
    'Prscrbr_Geo_Desc',
    'LA_Opioid_Prscrbng_Rate',        # Context
    'LA_Opioid_Prscrbng_Rate_1Y_Chg'  # TARGET
]

df_opioid = df_opioid[columns_to_keep]
df_opioid = df_opioid.rename(columns={
    'Prscrbr_Geo_Cd': 'FIPS_Code',
    'Prscrbr_Geo_Desc': 'County',
    'LA_Opioid_Prscrbng_Rate': 'LA_Opioid_Rate'
})
df_opioid
```

```python
df_opioid[["State","County"]] = df_opioid["County"].str.split(":", expand=True)
df_opioid
```

```python
df_opioid["FIPS_Code"] = df_opioid["FIPS_Code"].astype(int)
df_opioid
```

```python
master = df_opioid[[
    "Year",
    "FIPS_Code",
    "State",
    "County",
    "LA_Opioid_Rate",
    "LA_Opioid_Prscrbng_Rate_1Y_Chg"
]]
master
```

```python
education_data = education_data.drop(columns=["County"])
master = master.merge(
    education_data,
    on=["Year", "FIPS_Code"],
    how="left"
)
master
```

```python
unemployment_data = unemployment_data.drop(columns=["County"])
master = master.merge(
    unemployment_data,
    on=["Year", "FIPS_Code"],
    how="left"
)
master
```

```python
poverty_data = poverty_data.drop(columns=["County"])
master = master.merge(
    poverty_data,
    on=["Year", "FIPS_Code"],
    how="left"
)
master
```

```python
marital_data = marital_data.drop(columns=["County"])

master = master.merge(
    marital_data,
    on=["Year", "FIPS_Code"],
    how="left"
)
master
```

```python
master = master[[
    "Year",
    "FIPS_Code",
    "State",
    "County",
    "Labor_Force_Participation_Rate",
    "Unemployment_Rate",
    "Median_Household_Income",
    "Poverty_Percent_All_Ages",
    "Poverty_Percent_Age_0_17",
    "Pct_Never_Married",
    "Pct_Divorced",
    "Pct_Less_Than_HS",
    "Pct_HS_Grad",
    "Pct_Bachelors_Plus",
    "Rural_Urban_Continuum_Code_2023",
    "LA_Opioid_Rate",
    "LA_Opioid_Prscrbng_Rate_1Y_Chg"
]]
master
```

```python
master.to_csv("west_virginia_opioid_data.csv", index=False)
```
