```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from dotenv import load_dotenv
import os
```

```python
df = pd.read_csv("../data/west_virginia_opioid_data.csv", dtype={"FIPS_Code": str})
df
```

```python
print("Missing values BEFORE interpolation:")
print(df.isnull().sum())
print(f"\nTotal missing: {df.isnull().sum().sum()}")
```

```python
df = df.sort_values(["FIPS_Code", "Year"]).reset_index(drop=True)
```

```python
cols_to_interpolate = ["LA_Opioid_Rate", "LA_Opioid_Prscrbng_Rate_1Y_Chg"]
```

```python
df[cols_to_interpolate] = (
    df.groupby("FIPS_Code")[cols_to_interpolate]
    .transform(lambda group: group.interpolate(method="linear", limit_direction="both"))
)
```

```python
print("\nMissing values AFTER interpolation:")
print(df.isnull().sum())
print(f"\nTotal missing: {df.isnull().sum().sum()}")
```

```python
df.to_csv("../data/west_virginia_opioid_data_interpolated.csv", index=False)
```
