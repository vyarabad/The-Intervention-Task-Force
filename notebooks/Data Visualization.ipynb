```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests
from dotenv import load_dotenv
import os
```

```python
df = pd.read_csv("../data/west_virginia_opioid_data_interpolated.csv")
df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,6))
sns.lineplot(data=df, x="Year", y="LA_Opioid_Rate", hue="County", marker="o")
plt.title("Opioid Rate Over Time by County")
plt.ylabel("Opioid Rate per 1,000 Residents")
plt.xlabel("Year")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x="Poverty_Percent_All_Ages", y="LA_Opioid_Rate", hue="Year", palette="coolwarm")
plt.title("Opioid Rate vs Poverty Percent (All Ages)")
plt.xlabel("Poverty Percent (All Ages)")
plt.ylabel("Opioid Rate per 1,000 Residents")
plt.show()
```

```python
plt.figure(figsize=(10,8))
corr = df[[
    "LA_Opioid_Rate", "Poverty_Percent_All_Ages", "Median_Household_Income",
    "Pct_Never_Married", "Pct_Divorced", "Unemployment_Rate"
]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap: Socioeconomic Factors vs Opioid Rate")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(data=df, x="Rural_Urban_Continuum_Code_2023", y="LA_Opioid_Prscrbng_Rate_1Y_Chg")
plt.title("Opioid Rate by Rural-Urban Classification")
plt.xlabel("Rural-Urban Continuum Code (2023)")
plt.ylabel("Opioid Rate per 1,000 Residents")
plt.show()
```
