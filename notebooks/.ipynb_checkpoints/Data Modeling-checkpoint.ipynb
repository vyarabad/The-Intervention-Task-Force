```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
df = pd.read_csv("../data/west_virginia_opioid_data_clean.csv")
df
```

```python
print(df.columns.tolist())
```

```python
from sklearn.linear_model import LinearRegression

YEAR_COL = "Year"
TARGET = "LA_Opioid_Prscrbng_Rate_1Y_Chg"   # target

train_year = 2022
test_year = 2023

# Drop columns that are IDs or not usable as predictors
DROP_COLS = [YEAR_COL, TARGET, "FIPS_Code", "State", "County"]

FEATURES = [c for c in df.columns if c not in DROP_COLS]

print("Model: LinearRegression()")
print("Target:", TARGET)
print("Train/Test split:", train_year, "->", test_year)
print("Number of features:", len(FEATURES))
print("Features:", FEATURES)
```

```python
train_df = df[df[YEAR_COL] == train_year].copy()
test_df  = df[df[YEAR_COL] == test_year].copy()

train_df = train_df.dropna(subset=[TARGET])
test_df = test_df.dropna(subset=[TARGET])

X_train = train_df[FEATURES]
y_train = train_df[TARGET]

X_test = test_df[FEATURES]
y_test = test_df[TARGET]

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
import numpy as np

baseline_model = Pipeline([
    ("scaler", StandardScaler()),
    ("lr", LinearRegression())
])

baseline_model.fit(X_train, y_train)

y_pred = baseline_model.predict(X_test)

print("Done predicting. Example preds:", y_pred[:5])
```

```python
from sklearn.metrics import mean_squared_error, r2_score

rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("Baseline Linear Regression")
print("RMSE:", rmse)
print("R²:", r2)
```

```python
naive_pred = np.full(len(y_test), y_train.mean())

naive_rmse = np.sqrt(mean_squared_error(y_test, naive_pred))
naive_r2 = r2_score(y_test, naive_pred)

print("Naive Baseline (Predict Train Mean)")
print("RMSE:", naive_rmse)
print("R²:", naive_r2)
```

```python
import pandas as pd

results = pd.DataFrame({
    "Model": ["Linear Regression", "Naive Mean"],
    "RMSE": [rmse, naive_rmse],
    "R2": [r2, naive_r2]
})

results
```

```python
coefs = baseline_model.named_steps["lr"].coef_

coef_df = pd.DataFrame({
    "Feature": FEATURES,
    "Coefficient": coefs
})

coef_df["AbsCoeff"] = coef_df["Coefficient"].abs()

top5 = coef_df.sort_values("AbsCoeff", ascending=False).head(5)
top5
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.barh(top5["Feature"][::-1], top5["Coefficient"][::-1])
plt.title("Top 5 Predictive Features (Linear Regression Baseline)")
plt.xlabel("Coefficient Value")
plt.tight_layout()
plt.show()
```

```python
print("Do these drivers make sense?")
for f, c in zip(top5["Feature"], top5["Coefficient"]):
    direction = "increases" if c > 0 else "decreases"
    print(f"- {f}: higher values {direction} predicted 1-year change in prescribing rate")
```
